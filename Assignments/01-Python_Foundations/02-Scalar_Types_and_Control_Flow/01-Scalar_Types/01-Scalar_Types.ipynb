# 📘 01 - Scalar Types

Review and practice working with Python's basic scalar data types.

## ✅ Learning Goals
- Recognize scalar types (`int`, `float`, `str`, `bool`)
- Use the `type()` function to identify types
- Use variables and assignment

````python
# 🔧 Define some variables
num = 42
pi = 3.14159
name = 'Python'
is_cool = True
````

````python
# 🔍 Check their types
print(type(num))
print(type(pi))
print(type(name))
print(type(is_cool))
````

### ✏️ Your Turn
Create your own variables of each type and print them.

````python
age = 24
height = 5.7
name = "Manoj"
is_student = True

print(age)
print(height)
print(name)
print(is_student)

print(type(age))
print(type(height))
print(type(name))
print(type(is_student))
````

## ⚖️ Comparison Practice
Use comparison operators to check values.

````python
x = 10
y = 5
print(x > y)
print(x == y)
print(x != y)
````
