# 🧮 Sorting & Filtering in Pandas

## 🔹 LEARNING GOALS:
- Sort rows using `.sort_values()`
- Filter data using conditional logic
- Chain multiple filters together
- Use `.query()` for readable filtering


### 📥 1. Load the Dataset

```python
import pandas as pd

df = pd.read_csv("../../data/students.csv")
df.head()
```

### 🔢 2. Sorting Data

```python
# Sort by math_score
df.sort_values(by="MATH")
```

```python
# Sort by science_score descending
df.sort_values(by="SCIENCE", ascending=False)
```

```python
# Sort by multiple columns
df.sort_values(by=["SCIENCE", "MATH"], ascending=[False, True])
```

### 🧪 3. Filtering with Conditions

```python
# Who scored at least 90 in math?
df[df["MATH"] >= 90]
```

```python
# Who scored less than 70 in either subject?
df[(df["MATH"] < 70) | (df["SCIENCE"] < 70)]
```

### 🔗 4. Combining Filters

```python
# Students with B+ or higher in both
df[(df["MATH"] >= 87) & (df["SCIENCE"] >= 87)]
```

### 🧾 5. Using `.query()` for Readability

```python
# Same query using .query()
df.query("MATH >= 87 and SCIENCE >= 87")
```

### 🧪 Try It Yourself

- Sort the dataset by first name alphabetically
- Find students with the same math and science score
- Use `.query()` to find students who scored over 95 in *either* subject


```python
# Sort by first name alphabetically
print("Sorted by first name:")
print(df.sort_values(by="FIRST_NAME"))

# Find students with the same math and science score
print("\nSame math and science score:")
print(df[df["MATH"] == df["SCIENCE"]])

# Use query to find students who scored over 95 in either subject
print("\nScored over 95 in either subject:")
print(df.query("MATH > 95 or SCIENCE > 95"))
```

### 🧠 Mini-Challenge

> Load `"data/survey.csv"` and:
- Sort by the most recent signup date
- Filter out anyone who didn't complete the survey (nulls or "N/A")
- Use `.query()` to select rows with a specific condition of your choice


```python
import pandas as pd

# Load survey dataset
survey = pd.read_csv("../../data/survey.csv")

# Sort by most recent survey date
survey["survey_date"] = pd.to_datetime(
    survey["survey_date"],
    format="%Y-%m-%d",
    errors="coerce"
)

survey = survey.sort_values(by="survey_date", ascending=False)

# Replace N/A with missing values
survey = survey.replace("N/A", pd.NA)

# Remove incomplete rows
survey_clean = survey.dropna()

# Query students with GPA over 3.5
result = survey_clean.query("gpa > 3.5")

print("Cleaned Survey Data:")
print(survey_clean.head())

print("\nQuery Result:")
print(result.head())
```
