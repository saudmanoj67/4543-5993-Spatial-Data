```python
# 1. Load and inspect the data

import pandas as pd
import numpy as np

df = pd.read_csv("../../data/survey.csv")

print(df.head())

df.info()
```

```python
# 2. Clean up string fields

# Strip whitespace
string_cols = df.select_dtypes(include="object").columns

for col in string_cols:
    df[col] = df[col].astype(str).str.strip()

# Convert blank strings to NaN
df = df.replace(["", "N/A", "n/a", "None", "none"], np.nan)

print(df.head())
```

```python
# 3. missing values

print("Missing values before cleaning:")
print(df.isnull().sum())

# Fill numeric missing values with median
numeric_cols = df.select_dtypes(include="number").columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Fill text missing values with "Unknown"
text_cols = df.select_dtypes(include="object").columns

for col in text_cols:
    df[col] = df[col].fillna("Unknown")

print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

```python
# 4. Standardize data

if "gender" in df.columns:
    df["gender"] = df["gender"].str.lower().str.strip()

    df["gender"] = df["gender"].replace({
        "m": "Male",
        "male": "Male",
        "f": "Female",
        "female": "Female",
        "other": "Other"
    })

    print(df["gender"].value_counts())

if "exercise_freq" in df.columns:
    df["exercise_freq"] = df["exercise_freq"].str.lower().str.strip()

    df["exercise_freq"] = df["exercise_freq"].replace({
        "daily": "Daily",
        "weekly": "Weekly",
        "sometimes": "Sometimes",
        "rarely": "Rarely",
        "never": "Never"
    })

    print(df["exercise_freq"].value_counts())
```

```python
# 5. Fix data types

for col in ["age", "height_cm", "weight_kg", "satisfaction", "satisfaction_1_5"]:
    if col in df.columns:
        df[col] = (
            df[col]
            .astype(str)
            .str.extract(r"(\d+\.?\d*)")[0]
        )
        df[col] = pd.to_numeric(df[col], errors="coerce")

if "submitted_at" in df.columns:
    df["submitted_at"] = pd.to_datetime(df["submitted_at"], errors="coerce")

if "survey_date" in df.columns:
    df["survey_date"] = pd.to_datetime(
    df["survey_date"],
    format="%Y-%m-%d",
    errors="coerce"
)

df.info()
```

```python
# 6. Validate and clean emails

if "email" in df.columns:
    valid_email = df["email"].str.contains(
        r"^[\w\.-]+@[\w\.-]+\.\w+$",
        na=False
    )

    df.loc[~valid_email, "email"] = np.nan

    print("Invalid emails changed to NaN.")
    print(df["email"].isnull().sum())
else:
    print("No email column found in this dataset.")
```

```python
# 7. Convert consent to Boolean

if "consent_given" in df.columns:
    df["consent_given"] = df["consent_given"].astype(str).str.lower().str.strip()

    df["consent_given"] = df["consent_given"].replace({
        "yes": True,
        "y": True,
        "true": True,
        "1": True,
        "no": False,
        "n": False,
        "false": False,
        "0": False
    })

    print(df["consent_given"].value_counts())
else:
    print("No consent_given column found in this dataset.")
```

```python
# 8. Check for duplicates

print("Duplicate rows:")
print(df.duplicated().sum())

df = df.drop_duplicates()

if "student_id" in df.columns:
    print("Duplicate student IDs:")
    print(df["student_id"].duplicated().sum())

    df = df.drop_duplicates(subset="student_id")
```

```python
# 9. Save cleaned data

df.to_csv("../../data/survey_clean.csv", index=False)

print("survey_clean.csv saved successfully!")
```
