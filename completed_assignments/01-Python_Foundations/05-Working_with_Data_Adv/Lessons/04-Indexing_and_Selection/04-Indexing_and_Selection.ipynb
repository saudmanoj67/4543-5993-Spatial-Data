# 🎯 Notebook 04: Indexing and Selection

Time to slice and dice.

In this notebook, you'll:
- Use `.loc[]` for label-based selection
- Use `.iloc[]` for position-based selection
- Create boolean masks to filter rows
- Select specific columns and rows (solo or together)

Let the chaos of index manipulation begin.
---

```python
import pandas as pd

# Load Titanic CSV directly
df = pd.read_csv("../../data/titanic.csv")

df.head(3)
```

## 🔎 Select rows by label with `.loc[]`

```python
# .loc[row_label, column_label]
# Titanic uses default RangeIndex, so labels are integers
print(df.loc[0])

# Row range (inclusive!)
df.loc[0:4, ["Age", "Fare", "Sex"]]
```

## 🔢 Select rows by position with `.iloc[]`

```python
# iloc uses pure positions
df.iloc[0]  # First row

# Row slice, column slice
df.iloc[1:4, 0:5]
```

## 🎯 Boolean Masking (Filtering)

```python
# Filter passengers under age 10
kids = df[df["Age"] < 10]

kids[["Age", "Sex", "Pclass"]].head()
```

## 🧠 Compound Conditions

```python
# Adults in First Class who paid over $100
rich_folks = df[(df["Age"] > 18) & (df["Pclass"] == 1) &(df["Fare"] > 100)]

rich_folks[["Age", "Pclass", "Fare"]].head()
```

## 🧹 Select Columns Only

```python
df[["Sex", "Age", "Pclass"]].head()
```

## 🎯 Your Turn

1. Select the first row using `.iloc[]`, then again using `.loc[]`.
2. Slice the first 5 rows and select only `survived`, `sex`, and `age` columns.
3. Filter the dataset for female passengers over 40.
4. Create a DataFrame of passengers in 2nd class under age 15.
5. Use `.loc[]` to select the rows 10–20 and only the columns `fare`, `class`, and `age`.

🎯 **Bonus:** Find passengers with missing `age` and print only their `name`, `age`, and `pclass` (use `.isnull()`).

```python
# Your slicing magic goes here!

# First row using iloc
print("First row using iloc:")
print(df.iloc[0])

# First row using loc
print("\nFirst row using loc:")
print(df.loc[0])

# First 5 rows with selected columns
print("\nFirst 5 rows: Survived, Sex, Age")
print(df.loc[0:4, ["Survived", "Sex", "Age"]])

# Female passengers over 40
print("\nFemale passengers over 40:")
female_over_40 = df[(df["Sex"] == "female") & (df["Age"] > 40)]
print(female_over_40)

# Passengers in 2nd class under age 15
print("\n2nd class passengers under age 15:")
second_class_under_15 = df[(df["Pclass"] == 2) & (df["Age"] < 15)]
print(second_class_under_15)

# Rows 10-20 with Fare, Pclass, Age
print("\nRows 10-20 with Fare, Pclass, Age:")
print(df.loc[10:20, ["Fare", "Pclass", "Age"]])

# Bonus: missing age
print("\nPassengers with missing age:")
missing_age = df[df["Age"].isnull()]
print(missing_age[["Name", "Age", "Pclass"]])
```
