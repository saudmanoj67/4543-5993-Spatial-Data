# 🧼 Cleaning Dirty Data (Missing Values & Type Fixes)

## 🔹 LEARNING GOALS:
- Detect and count missing values (`NaN`)
- Fill or drop missing data
- Convert column data types safely
- Understand the difference between `NaN`, `None`, `""`, and type mismatches


### 🧪 1. Load a Messy Dataset

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", None, ""],
    "Age": ["25", "thirty", 35, np.nan, "40", "old"],
    "Signup Date": ["2022-01-01", "not a date", "2022/03/01", None, "April 5, 2022", "yesterday"],
    "Score": [95.5, None, 88.0, 92.5, "", "high"],
    "Email": ["alice@email.com", None, "", "david@email.com", None, ""]
}

df = pd.DataFrame(data)
df
```

### 🧯 2. Detecting Missing or Broken Values

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df[df.isnull().any(axis=1)]
```

### 🧹 3. Cleaning Strategy Options

```python
df.fillna({
    "Name": "Unknown",
    "Age": -1,
    "Signup Date": "1970-01-01",
    "Score": 0.0
})
```

```python
df.dropna()
```

### 🧬 4. Data Type Fixes

```python
df.dtypes
```

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Score"] = pd.to_numeric(df["Score"], errors="coerce")
df["Signup Date"] = pd.to_datetime(df["Signup Date"], errors="coerce")
df.dtypes
```

### 🩹 5. Impute (Fill In) Fixed Missing Values

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Score"] = df["Score"].fillna(df["Score"].mean())
```

```python
df["Signup Date"] = df["Signup Date"].fillna(df["Signup Date"].min())
```

```python
df["Name"] = df["Name"].fillna("Unknown")
```

```python
df["Email"] = df["Email"].replace("", np.nan)
df["Email"] = df["Email"].fillna("No Email")
```

### 🤓 6. Cleaned Data Review

```python
df.info()
```

```python
df.describe(include="all")
```

### 🧠 Mini-Challenge

> 🗂 Load `"data/survey.csv"` and:
- Identify which columns have missing values
- Use `.isnull().sum()` to get a null report
- Use a mix of `.fillna()`, `.dropna()`, and `pd.to_numeric()` or `pd.to_datetime()` to clean it
- Print a summary with `.info()` and `.describe()`

```python
import pandas as pd
import numpy as np

# Load survey dataset
df = pd.read_csv("../../data/survey.csv")

print("First 5 rows:")
print(df.head())

# Show missing values
print("\nMissing Values Report:")
print(df.isnull().sum())

# Replace empty strings with NaN
df = df.replace("", np.nan)

# Try converting numeric columns
for col in df.columns:
    try:
        df[col] = pd.to_numeric(df[col])
    except:
        pass

# Fill numeric missing values with mean
numeric_cols = df.select_dtypes(include="number").columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].mean())

# Drop rows that are completely empty
df = df.dropna(how="all")

# Final info
print("\nDataset Info:")
df.info()

# Describe dataset
print("\nSummary Statistics:")
print(df.describe(include="all"))
```

### 📝 Summary

| Concept        | Tool/Function                      |
|----------------|------------------------------------|
| Detect nulls   | `df.isnull()`, `df.isnull().sum()` |
| Drop rows      | `df.dropna()`                      |
| Fill values    | `df.fillna()`                      |
| Convert types  | `pd.to_numeric()`, `pd.to_datetime()` |
| Replace values | `df.replace()`                     |
